```python
# intersection of ids classified as non-hallu

targets_non_hallu = df[df["targets_hallu.1"] == False]["item_id"]
forecast_non_hallu = df[df["forecast_hallu.1"] == False]["item_id"]
```

```python
assert len(targets_non_hallu) == len(set(targets_non_hallu))
assert len(forecast_non_hallu) == len(set(forecast_non_hallu))

print(len(set(targets_non_hallu)), len(set(forecast_non_hallu)))
intersection = set(targets_non_hallu).intersection(set(forecast_non_hallu))
intersection.__len__() # the number of times both the target and the forecast were classified as non-hallucinating
```

```python
type3_targets_non_hallu = df[(df["target_ARMA rule.1"] == 1) & (df["target_pattern rule.1"] == 1)]["item_id"]
type3_forecasts_non_hallu = df[(df["forecast_ARMA rule.1"] == 1) & (df["forecast_pattern rule.1"] == 1)]["item_id"]
```

```python
print(type3_targets_non_hallu.shape)
print(type3_forecasts_non_hallu.shape)

intersection = set(type3_targets_non_hallu).intersection(set(type3_forecasts_non_hallu))
intersection.__len__() # the number of times both the target and the forecast were classified as non-hallucinating
```

```python
import pandas as pd

df = pd.read_csv("outputs/chronos_m4_context.csv")

df.head()
```

```python
ts_sizes = []
for id, row in df.iterrows():
    # print(row.isna().value_counts())
    false_count = row.notna().sum()
    ts_sizes.append(false_count)    
```

```python
ts_sizes = pd.Series(ts_sizes)
print(ts_sizes.mean(), ts_sizes.median(), ts_sizes.mode(), ts_sizes.std(), ts_sizes.var())
```
